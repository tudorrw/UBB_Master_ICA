```python
import pandas as pd
```

```python
df = pd.read_csv("datasets/global_volcano_eruption_intelligence.csv")
```

```python
# Volcano with the most eruption records in the dataset
volcano_counts = df["volcano_name"].value_counts()

print("\nTop 10 volcanoes by record count:")
print(volcano_counts.head(10))
```

```python
etna_df = df[df["volcano_name"] == "Etna"].copy().reset_index(drop=True)

# etna_df.to_csv("etna_df1.csv", index=False)
keep_numeric = {"noaa_event_id", "year", "month", "day"}
numeric_cols_to_drop = [
    col for col in etna_df.select_dtypes(include="number").columns
    if col not in keep_numeric
]
etna_df = etna_df.drop(columns=numeric_cols_to_drop)

# etna_df.to_csv("etna_df2.csv", index=False)
```

```python
from numpy import int8
remaining_etna_df = etna_df[["year", "month", "day", "has_casualties", "vei_category", "vei_risk_tier", "eruption_agent_codes", "eruption_agent_decoded"]].copy().reset_index(drop=True)


remaining_etna_df.dropna(how='any',axis=0, inplace=True) 
remaining_etna_df['month'] = remaining_etna_df['month'].astype(int8)

remaining_etna_df['month'] = pd.to_datetime(remaining_etna_df['month'], format='%m').dt.month_name()
remaining_etna_df.head(30)
```

```python
for col in remaining_etna_df.columns:
    # print(f"Attributes: {len(i.columns) - 1}\n")

    print(f"{col}: {remaining_etna_df[col].nunique()} values")
    print(remaining_etna_df[col].unique())
    
remaining_etna_df.drop(columns=["day"], inplace=True)
remaining_etna_df.drop(columns=["eruption_agent_decoded"], inplace=True)


```

```python
remaining_etna_df.info()
```

```python
MONTH_TO_SEASON = {
    "December": "Winter", "January": "Winter", "February": "Winter",
    "March": "Spring", "April": "Spring", "May": "Spring",
    "June": "Summer", "July": "Summer", "August": "Summer",
    "September": "Autumn", "October": "Autumn", "November": "Autumn",
}

# remaining_etna_df.insert(0, "id", [f"E{i}" for i in range(1, len(remaining_etna_df) + 1)])
remaining_etna_df["month"] = remaining_etna_df["month"].map(MONTH_TO_SEASON)

remaining_etna_df.rename(columns={'month': 'season', 'has_casualties': 'casualties', 'vei_category': 'category', 'vei_risk_tier': 'risk', 'eruption_agent_codes': 'agent'}, inplace=True)

remaining_etna_df.to_csv("datasets/etna_df.csv", index=False)
```

```python
time_part = remaining_etna_df[["year", "season"]]
time_part.to_csv("datasets/volcano_time_part.csv", index=False)

# time_part_int = remaining_etna_df[["id", "year", "season"]].rename(columns={"season": "month"})
# time_part_int.to_csv("datasets/volcano_time_part_int.csv", index=False)

event_part = remaining_etna_df.drop(columns=["year", "season"])
event_part.to_csv("datasets/volcano_event_part.csv", index=False)
```

```python
remaining_etna_df = remaining_etna_df.reset_index(drop=True)
remaining_etna_df.insert(0, "id", range(1, len(remaining_etna_df) + 1))

remaining_etna_df.to_csv("datasets/etna_df.csv", index=False)
remaining_etna_df[["id", "year", "season"]].to_csv("datasets/volcano_time_part.csv", index=False)
remaining_etna_df.drop(columns=["year", "season"]).to_csv("datasets/volcano_event_part.csv", index=False)

remaining_etna_df
```
