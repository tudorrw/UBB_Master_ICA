```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/llm_benchmark_comparison_2025_2026.csv")
df.sample(5)

```

```python
benchmark_df = df[["model_name", "mmlu", "swe_bench", "humaneval"]].copy()

# benchmark_df.to_csv("benchmark_df.csv", index=False)

def map_mmlu_to_tier(mmlu_score):
    if mmlu_score >= 90.0:
        return 'High'
    elif mmlu_score >= 85.0:
        return 'Med'
    else:
        return 'Low'

def map_swe_bench_to_tier(swe_bench_score):
    if swe_bench_score >= 75.0:
        return 'High'
    elif swe_bench_score >= 55.0:
        return 'Med'
    else:
        return 'Low'

def map_humaneval_to_tier(humaneval_score):
    if humaneval_score >= 94.0:
        return 'High'
    elif humaneval_score >= 90.0:
        return 'Med'
    else:
        return 'Low'
benchmark_df['mmlu_tier'] = benchmark_df['mmlu'].apply(map_mmlu_to_tier)
benchmark_df['swe_bench_tier'] = benchmark_df['swe_bench'].apply(map_swe_bench_to_tier)
benchmark_df['humaneval_tier'] = benchmark_df['humaneval'].apply(map_humaneval_to_tier)
# Use .isin() instead of 'in'
# filtered_df = benchmark_df[benchmark_df['model_name'].isin(['Claude Opus 4.5', 'Gemini 2.5 Pro', 'MiniMax-M2.5', 'DeepSeek R1', 'Mistral Large 3'])]

benchmark_df.head(24)
```
